```python
# ==============================================================================
# PARTE A: ANÁLISIS EXPLORATORIO - MONITOREO DE IMPEDANCIA BIOELÉCTRICA
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt

# 1. Ingreso de los 30 datos experimentales (Frecuencia en Hz, Impedancia en Ohm)
# Los datos corresponden al experimento controlado a 37°C.
frecuencias = np.array([
    100, 120, 145, 170, 200, 235, 270, 310, 355, 405,
    460, 520, 585, 655, 730, 810, 895, 985, 1080, 1180,
    1290, 1410, 1540, 1680, 1830, 1990, 2160, 2340, 2530, 2730
])

impedancias = np.array([
    152.3, 149.1, 146.8, 144.9, 142.0, 139.5, 137.9, 136.1, 134.8, 133.6,
    132.7, 131.9, 131.4, 131.1, 130.9, 131.0, 131.3, 131.9, 132.7, 133.8,
    135.2, 136.9, 138.9, 141.1, 143.5, 146.1, 149.0, 152.2, 155.6, 159.2
])

# 2. Configuración de la gráfica
plt.figure(figsize=(10, 6), dpi=100)

# Dibujar los puntos medidos en el laboratorio
plt.scatter(frecuencias, impedancias, color='#e74c3c', s=40, edgecolors='black', label='Datos medidos (f_i, |Z_i|)', zorder=5)

# Dibujar una línea discontinua suave para guiar el análisis visual de la tendencia
plt.plot(frecuencias, impedancias, linestyle='--', color='#2c3e50', alpha=0.5, label='Tendencia visual')

# Destacar visualmente el mínimo estimado de la tabla (Muestra 15: f = 730 Hz, |Z| = 130.9 Ohm)
plt.axvline(x=730, color='#27ae60', linestyle=':', linewidth=1.5, label='Mínimo visual estimado (~730 Hz)')
plt.scatter(730, 130.9, color='#27ae60', s=100, marker='*', edgecolors='black', zorder=6, label='Punto mínimo')

# 3. Personalización de etiquetas y títulos
plt.title('Análisis Exploratorio: Magnitud de Impedancia |Z| vs Frecuencia (f)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Frecuencia f (Hz)', fontsize=12)
plt.ylabel('Magnitud de Impedancia |Z| (Ohms)', fontsize=12)

# Añadir cuadrícula (grid) para facilitar la lectura de coordenadas
plt.grid(True, linestyle=':', alpha=0.6, color='gray')

# Ubicación de la leyenda
plt.legend(loc='upper center', frameon=True, facecolor='white', edgecolor='none', shadow=True)

# 4. Mostrar el gráfico en pantalla
plt.show()
```
