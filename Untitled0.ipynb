```python
# ==============================================================================
# PARTE A: ANÁLISIS EXPLORATORIO - MONITOREO DE IMPEDANCIA BIOELÉCTRICA
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt

# 1. Ingreso de los 30 datos experimentales (Frecuencia en Hz, Impedancia en Ohm)
# Los datos corresponden al experimento controlado a 37°C.
frecuencias = np.array([
    100, 120, 145, 170, 200, 235, 270, 310, 355, 405,
    460, 520, 585, 655, 730, 810, 895, 985, 1080, 1180,
    1290, 1410, 1540, 1680, 1830, 1990, 2160, 2340, 2530, 2730
])

impedancias = np.array([
    152.3, 149.1, 146.8, 144.9, 142.0, 139.5, 137.9, 136.1, 134.8, 133.6,
    132.7, 131.9, 131.4, 131.1, 130.9, 131.0, 131.3, 131.9, 132.7, 133.8,
    135.2, 136.9, 138.9, 141.1, 143.5, 146.1, 149.0, 152.2, 155.6, 159.2
])

# 2. Configuración de la gráfica
plt.figure(figsize=(10, 6), dpi=100)

# Dibujar los puntos medidos en el laboratorio
plt.scatter(frecuencias, impedancias, color='#e74c3c', s=40, edgecolors='black', label='Datos medidos (f_i, |Z_i|)', zorder=5)

# Dibujar una línea discontinua suave para guiar el análisis visual de la tendencia
plt.plot(frecuencias, impedancias, linestyle='--', color='#2c3e50', alpha=0.5, label='Tendencia visual')

# Destacar visualmente el mínimo estimado de la tabla (Muestra 15: f = 730 Hz, |Z| = 130.9 Ohm)
plt.axvline(x=730, color='#27ae60', linestyle=':', linewidth=1.5, label='Mínimo visual estimado (~730 Hz)')
plt.scatter(730, 130.9, color='#27ae60', s=100, marker='*', edgecolors='black', zorder=6, label='Punto mínimo')

# 3. Personalización de etiquetas y títulos
plt.title('Análisis Exploratorio: Magnitud de Impedancia |Z| vs Frecuencia (f)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Frecuencia f (Hz)', fontsize=12)
plt.ylabel('Magnitud de Impedancia |Z| (Ohms)', fontsize=12)

# Añadir cuadrícula (grid) para facilitar la lectura de coordenadas
plt.grid(True, linestyle=':', alpha=0.6, color='gray')

# Ubicación de la leyenda
plt.legend(loc='upper center', frameon=True, facecolor='white', edgecolor='none', shadow=True)

# 4. Mostrar el gráfico en pantalla
plt.show()
```

```python
# ==============================================================================
# PARTE B1: INTERPOLACIÓN POLINÓMICA Y FENÓMENO DE RUNGE
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial.polynomial import Polynomial

# Datos (asegurándonos de tenerlos en esta celda también)
frecuencias = np.array([100, 120, 145, 170, 200, 235, 270, 310, 355, 405, 460, 520, 585, 655, 730, 810, 895, 985, 1080, 1180, 1290, 1410, 1540, 1680, 1830, 1990, 2160, 2340, 2530, 2730])
impedancias = np.array([152.3, 149.1, 146.8, 144.9, 142.0, 139.5, 137.9, 136.1, 134.8, 133.6, 132.7, 131.9, 131.4, 131.1, 130.9, 131.0, 131.3, 131.9, 132.7, 133.8, 135.2, 136.9, 138.9, 141.1, 143.5, 146.1, 149.0, 152.2, 155.6, 159.2])

# Crear un vector de 500 puntos para ver las curvas suaves y las oscilaciones
f_fina = np.linspace(100, 2730, 500)

# Configuración del gráfico
plt.figure(figsize=(12, 7))
plt.scatter(frecuencias, impedancias, color='black', label='Datos reales', zorder=10)

# Grados a evaluar según el enunciado
grados = [5, 10, 15, 29]
colores = ['blue', 'green', 'orange', 'red']
nombres = ['Grado 5', 'Grado 10', 'Grado 15', 'Grado 29 (Interpolación exacta)']

# NOTA: Usamos np.polynomial.Polynomial.fit porque hace un escalado interno (método matricial estabilizado).
# Si usáramos matrices de Vandermonde puras o Lagrange sin escalar, el grado 29 colapsaría la memoria de la PC.
for g, c, nom in zip(grados, colores, nombres):
    poly = Polynomial.fit(frecuencias, impedancias, g)
    Z_estimada = poly(f_fina)
    plt.plot(f_fina, Z_estimada, color=c, alpha=0.7, label=nom, linewidth=2 if g==29 else 1.5)

# Limitamos el eje Y porque el grado 29 oscilará al infinito y romperá la escala visual
plt.ylim(120, 170)
plt.title('Evidencia del Fenómeno de Runge en Polinomios de Alto Grado')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Impedancia |Z| (Ohms)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# ==============================================================================
# PARTE B1 (Continuación): PREDICCIÓN Y VALIDACIÓN LEAVE-ONE-OUT (LOO)
# ==============================================================================

# 1. Cálculo del valor a 1000 Hz con el polinomio de Grado 5 (El más estable)
grado_seleccionado = 5
poly_5 = Polynomial.fit(frecuencias, impedancias, grado_seleccionado)
Z_1000 = poly_5(1000)

print("--- 1. PREDICCIÓN A 1000 Hz ---")
print(f"La impedancia interpolada a 1000 Hz (Polinomio Grado 5) es: {Z_1000:.4f} Ohms\n")

# 2. Validación Leave-One-Out (LOO) con 5 puntos al azar
print("--- 2. VALIDACIÓN LEAVE-ONE-OUT (LOO) ---")
np.random.seed(42) # Fijamos semilla para que el resultado sea reproducible
indices_azar = np.random.choice(len(frecuencias), 5, replace=False)

errores_relativos = []

for idx in indices_azar:
    f_oculto = frecuencias[idx]
    Z_real = impedancias[idx]

    # "Esconder" el punto seleccionado
    f_entrenamiento = np.delete(frecuencias, idx)
    Z_entrenamiento = np.delete(impedancias, idx)

    # Ajustar el modelo usando solo los 29 puntos restantes
    poly_loo = Polynomial.fit(f_entrenamiento, Z_entrenamiento, grado_seleccionado)

    # Predecir el valor en la frecuencia oculta
    Z_predicho = poly_loo(f_oculto)

    # Calcular error relativo
    error = abs(Z_real - Z_predicho) / Z_real
    errores_relativos.append(error)

    print(f"Punto oculto: f = {f_oculto} Hz | Z real = {Z_real} Ohm | Z predicho = {Z_predicho:.4f} Ohm | Error = {error*100:.4f}%")

# 3. Error promedio
error_promedio = np.mean(errores_relativos)
print(f"\n=> Error Relativo Promedio del modelo (LOO 5 puntos): {error_promedio*100:.4f}%")
```

```python
# ==============================================================================
# PARTE B2: SPLINES CÚBICOS NATURALES VS POLINOMIO DE GRADO 5
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
from numpy.polynomial.polynomial import Polynomial

# 1. Cargar Datos (los repetimos por si reiniciaste Colab)
frecuencias = np.array([100, 120, 145, 170, 200, 235, 270, 310, 355, 405, 460, 520, 585, 655, 730, 810, 895, 985, 1080, 1180, 1290, 1410, 1540, 1680, 1830, 1990, 2160, 2340, 2530, 2730])
impedancias = np.array([152.3, 149.1, 146.8, 144.9, 142.0, 139.5, 137.9, 136.1, 134.8, 133.6, 132.7, 131.9, 131.4, 131.1, 130.9, 131.0, 131.3, 131.9, 132.7, 133.8, 135.2, 136.9, 138.9, 141.1, 143.5, 146.1, 149.0, 152.2, 155.6, 159.2])

# 2. Recreamos el polinomio de Grado 5 de la parte B1
poly_5 = Polynomial.fit(frecuencias, impedancias, 5)

# 3. Construimos el Spline Cúbico Natural (bc_type='natural' hace que la 2da derivada en los extremos sea 0)
spline_natural = CubicSpline(frecuencias, impedancias, bc_type='natural')

# 4. Evaluamos a 1000 Hz
Z_1000_poly = poly_5(1000)
Z_1000_spline = spline_natural(1000)

print(f"--- PREDICCIÓN A 1000 Hz ---")
print(f"Impedancia con Polinomio (Grado 5): {Z_1000_poly:.4f} Ohms")
print(f"Impedancia con Spline Cúbico:       {Z_1000_spline:.4f} Ohms")

# 5. Malla fina para graficar y comparar ambas curvas
f_fina = np.linspace(100, 2730, 1000)
Z_poly_fina = poly_5(f_fina)
Z_spline_fina = spline_natural(f_fina)

# 6. Gráfica Comparativa
plt.figure(figsize=(10, 6))
plt.scatter(frecuencias, impedancias, color='black', label='Datos medidos', zorder=5)

# Gráfica del Spline (suave y pasa por todos los puntos)
plt.plot(f_fina, Z_spline_fina, color='blue', linewidth=2, label='Spline Cúbico Natural')

# Gráfica del Polinomio (suave, pero es una curva de tendencia global)
plt.plot(f_fina, Z_poly_fina, color='red', linestyle='--', linewidth=2, label='Polinomio Grado 5')

# Marcador en 1000 Hz
plt.axvline(x=1000, color='gray', linestyle=':', label='Frecuencia = 1000 Hz')
plt.scatter([1000, 1000], [Z_1000_spline, Z_1000_poly], color=['blue', 'red'], zorder=6, s=60)

plt.title('Comparación: Spline Cúbico Natural vs Polinomio (Grado 5)')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Impedancia |Z| (Ohms)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# ==============================================================================
# PARTE C: DERIVACIÓN NUMÉRICA MEDIANTE SPLINE CÚBICO ANALÍTICO
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
from scipy.optimize import brentq

# 1. Cargar Datos
frecuencias = np.array([100, 120, 145, 170, 200, 235, 270, 310, 355, 405, 460, 520, 585, 655, 730, 810, 895, 985, 1080, 1180, 1290, 1410, 1540, 1680, 1830, 1990, 2160, 2340, 2530, 2730])
impedancias = np.array([152.3, 149.1, 146.8, 144.9, 142.0, 139.5, 137.9, 136.1, 134.8, 133.6, 132.7, 131.9, 131.4, 131.1, 130.9, 131.0, 131.3, 131.9, 132.7, 133.8, 135.2, 136.9, 138.9, 141.1, 143.5, 146.1, 149.0, 152.2, 155.6, 159.2])

# 2. Construcción del Spline Cúbico Natural
spline_natural = CubicSpline(frecuencias, impedancias, bc_type='natural')

# 3. Derivación analítica del Spline (nu=1 para primera derivada, nu=2 para la segunda)
dspline_1 = spline_natural.derivative(nu=1)
dspline_2 = spline_natural.derivative(nu=2)

# 4. Encontrar la raíz exacta de la primera derivada (donde d|Z|/df = 0)
# Observamos que cambia de signo entre 730 Hz (negativa) y 810 Hz (positiva)
f_min_exacta = brentq(dspline_1, 730, 810)
z_min_exacto = spline_natural(f_min_exacta)
d2z_en_minimo = dspline_2(f_min_exacta)

print("--- RESULTADOS MATEMÁTICOS DE ALTA PRECISIÓN ---")
print(f"Ubicación exacta del mínimo (f):        {f_min_exacta:.4f} Hz")
print(f"Impedancia mínima calculada (|Z|):     {z_min_exacto:.4f} Ohms")
print(f"Segunda derivada en el mínimo (d²/df²): {d2z_en_minimo:.8f} Ohms/Hz²\n")

# 5. Malla fina para graficar la primera derivada continua
f_fina = np.linspace(100, 2730, 1000)
derivada_fina = dspline_1(f_fina)

# 6. Gráfica de la Primera Derivada d|Z|/df vs f
plt.figure(figsize=(11, 5))
plt.plot(f_fina, derivada_fina, color='#8e44ad', linewidth=2, label='Primera Derivada d|Z|/df (Analítica del Spline)')
plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)

# Destacar el punto donde cruza por cero (el mínimo)
plt.vlines(x=f_min_exacta, ymin=min(derivada_fina), ymax=max(derivada_fina), colors='green', linestyles=':', alpha=0.7)
plt.scatter(f_min_exacta, 0, color='green', s=80, zorder=5, label=f'Cruce por cero (Mínimo a {f_min_exacta:.2f} Hz)')

plt.title('Primera Derivada de la Impedancia d|Z|/df vs Frecuencia', fontsize=13, fontweight='bold')
plt.xlabel('Frecuencia f (Hz)', fontsize=11)
plt.ylabel('d|Z|/df (Ohms / Hz)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
